# Basic Signal Concepts

Fundamental concepts of sinusoidal signals, including amplitude, frequency, period, phase, RMS, and power.

## Objective

Understand the basic properties of a sinusoidal signal and observe how changing each parameter affects the waveform.

We will investigate:

- Amplitude
- Frequency
- Period
- Phase
- RMS
- Power

## Theory

A sinusoidal signal can be written as:

$$
x(t) = A\sin(2\pi f t + \phi)
$$

where:

- $A$ = amplitude
- $f$ = frequency
- $T = 1/f$ = period
- $\phi$ = phase

For a sinusoidal signal:

$$
X_{RMS} = \frac{A}{\sqrt{2}}
$$

and, for a 1 Ω load:

$$
P = X_{RMS}^2
$$

```python
import numpy as np
import matplotlib.pyplot as plt

fs = 1000
duration = 1.0

t = np.arange(0, duration, 1/fs)

A = 2
f = 10
phi = 0

x = A * np.sin(2 * np.pi * f * t + phi)
```

```python
plt.figure(figsize=(10, 4))

plt.plot(t, x)
plt.xlabel("Time (s)")
plt.ylabel("Amplitude")
plt.title("Sinusoidal Signal")
plt.grid(True)

plt.show()
```

```python
amplitudes = [0.5, 1, 2, 4]

plt.figure(figsize=(10, 5))

for A in amplitudes:
    x = A * np.sin(2 * np.pi * f * t)
    plt.plot(t[:200], x[:200], label=f"A={A}")

plt.xlabel("Time (s)")
plt.ylabel("Amplitude")
plt.title("Effect of Amplitude")
plt.legend()
plt.grid(True)

plt.show()
```

### Observation

Increasing the amplitude increases the signal magnitude but does not change its frequency.

```python
frequencies = [5, 10, 20, 40]

plt.figure(figsize=(10, 6))

for f in frequencies:
    x = np.sin(2 * np.pi * f * t)
    plt.plot(t[:200], x[:200], label=f"f={f} Hz")

plt.xlabel("Time (s)")
plt.ylabel("Amplitude")
plt.title("Effect of Frequency")
plt.legend()
plt.grid(True)

plt.show()
```

### Observation

Increasing the frequency increases the number of oscillations within the same time interval.

```python
phases = [0, np.pi/4, np.pi/2, np.pi]

plt.figure(figsize=(10, 6))

for phi in phases:
    x = np.sin(2 * np.pi * 10 * t + phi)
    plt.plot(t[:200], x[:200], label=f"φ={phi:.2f} rad")

plt.xlabel("Time (s)")
plt.ylabel("Amplitude")
plt.title("Effect of Phase")
plt.legend()
plt.grid(True)

plt.show()
```

### Observation

Changing the phase shifts the position of the waveform relative to the time reference.

```python
A = 2
f = 10

x = A * np.sin(2 * np.pi * f * t)

rms = np.sqrt(np.mean(x**2))
power = rms**2

print(f"RMS: {rms:.4f}")
print(f"Power (1 Ω): {power:.4f}")
```

```python
theoretical_rms = A / np.sqrt(2)
theoretical_power = theoretical_rms**2

print(f"Theoretical RMS: {theoretical_rms:.4f}")
print(f"Theoretical Power: {theoretical_power:.4f}")
```

## Observations

- Increasing amplitude changes the signal magnitude but not its frequency.
- Increasing frequency makes the waveform oscillate faster.
- Changing phase shifts the waveform in time.
- The measured RMS value is close to the theoretical value $A/\sqrt{2}$.
- For a 1 Ω load, the calculated power is approximately equal to $V_{RMS}^2$.

## Key Takeaways

- Amplitude controls signal magnitude.
- Frequency determines how fast the signal oscillates.
- Period is the inverse of frequency.
- Phase determines the signal's position relative to a reference.
- RMS provides an effective measure of sinusoidal signal magnitude.
- Power depends on RMS value and load resistance.
